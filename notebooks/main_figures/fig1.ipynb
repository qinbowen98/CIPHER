### Figure 1c — Modality overlap as an UpSet plot


```python
import os
import itertools
import logging
import warnings
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')
logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['svg.fonttype'] = 'none'
mpl.rcParams['font.family'] = 'DejaVu Sans'
BASE = '.'
OUT_DIR = os.path.join(BASE, 'figs1_final', 'CHD_trainset_four_modal_sample_upset')
os.makedirs(OUT_DIR, exist_ok=True)
ENDPOINT = 'CHD'
ENDPOINT_LABEL = 'CHD'
RISK_MODELS = {'SNP': 'code_02_02_save/snp_20260202_1432/preds/{endpoint}_snp_TR_train_best_pred.csv', 'Soma': 'code_02_02_save/soma_20260202_2006/preds/{endpoint}_soma_TR_train_best_pred.csv', 'ECG': 'code_02_02_save/ecg_20260203_1003/preds/{endpoint}_ecg_transformer_h256_l3_train_best_pred.csv', 'Fundus': 'code_02_02_save/fundus_20251229_1519/preds/{endpoint}_fundus_vitsmallFT_train_best_pred.csv'}
MODALITY_ORDER = ['SNP', 'Soma', 'ECG', 'Fundus']
MM_TO_INCH = 1 / 25.4
FIG_W_MM = 145
FIG_H_MM = 62
FIGSIZE = (FIG_W_MM * MM_TO_INCH, FIG_H_MM * MM_TO_INCH)
BAR_COLOR = 'black'
DOT_COLOR = 'black'
EMPTY_DOT_FACE = 'white'
LINE_COLOR = 'black'

def check_file(path):
    if not os.path.exists(str(OUTPUT_ROOT / path if (OUTPUT_ROOT / path).exists() else INPUT_ROOT / path)):
        raise FileNotFoundError(path)

def all_combinations_ordered():
    combos = []
    for degree in [4, 3, 2, 1]:
        for comb in itertools.combinations(MODALITY_ORDER, degree):
            combos.append(tuple(comb))
    return combos

def load_risk_pids(endpoint, modality):
    risk_fp = os.path.join(BASE, RISK_MODELS[modality].format(endpoint=endpoint))
    check_file(risk_fp)
    df = pd.read_csv(str(OUTPUT_ROOT / risk_fp if (OUTPUT_ROOT / risk_fp).exists() else INPUT_ROOT / risk_fp))
    if 'ID' in df.columns:
        df = df.rename(columns={'ID': 'pid'})
    if 'pid' not in df.columns:
        raise ValueError(f'No pid column in risk file: {risk_fp}. Available columns: {list(df.columns)}')
    df['pid'] = pd.to_numeric(df['pid'], errors='coerce')
    df = df.dropna(subset=['pid']).copy()
    df['pid'] = df['pid'].astype(int)
    pids = set(df['pid'].unique().tolist())
    print(f'[{endpoint} | {modality}] train samples: {len(pids)}')
    return pids

def build_sample_presence(endpoint):
    modality_to_pids = {}
    for modality in MODALITY_ORDER:
        modality_to_pids[modality] = load_risk_pids(endpoint, modality)
    modality_totals = {modality: len(modality_to_pids[modality]) for modality in MODALITY_ORDER}
    all_pids = sorted(set.union(*[modality_to_pids[m] for m in MODALITY_ORDER]))
    rows = []
    for pid in all_pids:
        present = [m for m in MODALITY_ORDER if pid in modality_to_pids[m]]
        rows.append({'endpoint': endpoint, 'pid': pid, 'n_modalities': len(present), 'signature': '|'.join(present), **{m: int(m in present) for m in MODALITY_ORDER}})
    sample_df = pd.DataFrame(rows)
    out_csv = os.path.join(OUT_DIR, f'{endpoint}_train_sample_modality_presence.csv')
    sample_df.to_csv(out_csv, index=False)
    print('Saved sample presence table:', out_csv)
    modality_total_df = pd.DataFrame({'modality': list(modality_totals.keys()), 'n_samples': list(modality_totals.values())})
    out_total_csv = os.path.join(OUT_DIR, f'{endpoint}_train_modality_total_counts.csv')
    modality_total_df.to_csv(out_total_csv, index=False)
    print('Saved modality total counts:', out_total_csv)
    return (sample_df, modality_totals)

def build_upset_counts(sample_df, endpoint):
    rows = []
    for comb in all_combinations_ordered():
        comb_txt = '|'.join(comb)
        mask = np.ones(len(sample_df), dtype=bool)
        for modality in MODALITY_ORDER:
            if modality in comb:
                mask &= sample_df[modality].values == 1
            else:
                mask &= sample_df[modality].values == 0
        count = int(sample_df.loc[mask, 'pid'].nunique())
        rows.append({'endpoint': endpoint, 'combination': comb_txt, 'degree': len(comb), 'count': count})
    count_df = pd.DataFrame(rows)
    out_csv = os.path.join(OUT_DIR, f'{endpoint}_train_sample_upset_counts_exact.csv')
    count_df.to_csv(out_csv, index=False)
    print('Saved UpSet counts:', out_csv)
    four_modal_n = int(count_df.loc[count_df['combination'] == '|'.join(MODALITY_ORDER), 'count'].sum())
    print(f'[{endpoint}] four-modal overlap n = {four_modal_n}')
    return count_df

def plot_upset(endpoint, count_df, modality_totals):
    fig = plt.figure(figsize=FIGSIZE)
    gs = fig.add_gridspec(2, 1, height_ratios=[2.55, 1.2], hspace=0.04)
    ax_bar = fig.add_subplot(gs[0, 0])
    ax_dot = fig.add_subplot(gs[1, 0])
    count_df = count_df[count_df['count'] > 0].copy()
    if len(count_df) == 0:
        ax_bar.text(0.5, 0.5, 'No samples', ha='center', va='center', fontsize=8)
        ax_bar.axis('off')
        ax_dot.axis('off')
        return
    x = np.arange(len(count_df))
    counts = count_df['count'].values
    combos = count_df['combination'].tolist()
    ax_bar.bar(x, counts, color=BAR_COLOR, width=0.72, linewidth=0)
    ymax = np.max(counts) * 1.33
    if ymax < 10:
        ymax = 10
    ax_bar.set_ylim(0, ymax)
    full_combo = '|'.join(MODALITY_ORDER)
    for (xi, c, comb) in zip(x, counts, combos):
        label = str(c)
        if comb == full_combo:
            label = f'4-modal\nn={c}'
        ax_bar.text(xi, c + ymax * 0.02, label, ha='center', va='bottom', fontsize=6.2, fontweight='bold' if comb == full_combo else 'normal')
    ax_bar.set_title(f'{ENDPOINT_LABEL} train-set sample overlap', fontsize=9, pad=4)
    ax_bar.set_ylabel('N samples', fontsize=7)
    ax_bar.set_xticks([])
    ax_bar.tick_params(axis='y', labelsize=6.5, length=2)
    for s in ['top', 'right']:
        ax_bar.spines[s].set_visible(False)
    ax_bar.spines['left'].set_linewidth(0.8)
    ax_bar.spines['bottom'].set_linewidth(0.8)
    total_text = f"SNP: {modality_totals['SNP']}    Soma: {modality_totals['Soma']}    ECG: {modality_totals['ECG']}    Fundus: {modality_totals['Fundus']}"
    ax_bar.text(0.995, 0.98, total_text, transform=ax_bar.transAxes, ha='right', va='top', fontsize=6.8, bbox=dict(facecolor='white', edgecolor='black', linewidth=0.5, boxstyle='round,pad=0.22'))
    y_pos = {m: i for (i, m) in enumerate(MODALITY_ORDER)}
    ax_dot.set_xlim(-0.5, len(count_df) - 0.5)
    ax_dot.set_ylim(-0.5, len(MODALITY_ORDER) - 0.5)
    ax_dot.invert_yaxis()
    for (xi, comb_txt) in zip(x, combos):
        comb = comb_txt.split('|')
        active_y = [y_pos[m] for m in comb]
        for modality in MODALITY_ORDER:
            ax_dot.scatter(xi, y_pos[modality], s=24, facecolor=EMPTY_DOT_FACE, edgecolor=DOT_COLOR, linewidth=0.6, zorder=1)
        if len(active_y) >= 2:
            ax_dot.plot([xi, xi], [min(active_y), max(active_y)], color=LINE_COLOR, linewidth=0.9, zorder=2)
        for modality in comb:
            ax_dot.scatter(xi, y_pos[modality], s=26, facecolor=DOT_COLOR, edgecolor=DOT_COLOR, linewidth=0.6, zorder=3)
    ax_dot.set_yticks(list(y_pos.values()))
    ax_dot.set_yticklabels(MODALITY_ORDER, fontsize=7.2)
    ax_dot.set_xticks([])
    ax_dot.tick_params(length=0)
    for s in ['top', 'right', 'left', 'bottom']:
        ax_dot.spines[s].set_visible(False)
    plt.tight_layout()
    out_pdf = os.path.join(OUT_DIR, f'{endpoint}_train_sample_upset_bw_wide.pdf')
    out_svg = out_pdf.replace('.pdf', '.svg')
    plt.savefig(out_pdf, dpi=300, transparent=True, bbox_inches='tight')
    plt.savefig(out_svg, dpi=300, transparent=True, bbox_inches='tight')
    plt.close()
    print('Saved UpSet plot:', out_pdf)

def main():
    print('\n====================================')
    print(ENDPOINT)
    print('====================================')
    (sample_df, modality_totals) = build_sample_presence(ENDPOINT)
    count_df = build_upset_counts(sample_df, ENDPOINT)
    plot_upset(ENDPOINT, count_df, modality_totals)
    print('\nAll done.')
    print('Output:', OUT_DIR)
if __name__ == '__main__':
    main()
```

### Figure 1c — BD modality overlap


```python
import os
import itertools
import logging
import warnings
from pathlib import Path
from collections import OrderedDict
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
try:
    from venn import venn
except ImportError:
    raise ImportError('Please install four-set venn package first:\npip install venn')
warnings.filterwarnings('ignore')
logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['svg.fonttype'] = 'none'
mpl.rcParams['font.family'] = 'DejaVu Sans'
BASE = Path('.')
OUT_DIR = BASE / 'fig1_final' / 'CMCS_trainset_four_modal_sample_4set_venn_forced_totals_ordered'
OUT_DIR.mkdir(exist_ok=True, parents=True)
REGION_DIR = OUT_DIR / 'exact_region_tables'
REGION_DIR.mkdir(exist_ok=True, parents=True)
ENDPOINT = 'CHD'
ENDPOINT_LABEL = 'CMCS'
RISK_MODELS = {'SNP': 'code_02_02_save/snp_20260202_1432/preds/{endpoint}_snp_TR_train_best_pred.csv', 'ECG': 'code_02_02_save/ecg_20260203_1003/preds/{endpoint}_ecg_transformer_h256_l3_train_best_pred.csv', 'Soma': 'code_02_02_save/soma_20260202_2006/preds/{endpoint}_soma_TR_train_best_pred.csv', 'Fundus': 'code_02_02_save/fundus_20251229_1519/preds/{endpoint}_fundus_vitsmallFT_train_best_pred.csv'}
MODALITY_ORDER = ['SNP', 'ECG', 'Soma', 'Fundus']
DISPLAY_LABELS = {'SNP': 'SNP', 'ECG': 'ECG', 'Soma': 'Plasma proteomics', 'Fundus': 'Fundus'}
MM_TO_INCH = 1 / 25.4
FIG_W_MM = 180
FIG_H_MM = 125
FIGSIZE = (FIG_W_MM * MM_TO_INCH, FIG_H_MM * MM_TO_INCH)
LEGEND_W_MM = 95
LEGEND_H_MM = 42
LEGEND_FIGSIZE = (LEGEND_W_MM * MM_TO_INCH, LEGEND_H_MM * MM_TO_INCH)
VENN_COLORS = {'Fundus': '#639ebe', 'ECG': '#1a6292', 'Soma': '#af7bac', 'SNP': '#7d5398'}
VENN_COLORS = {'Fundus': '#778368', 'ECG': '#658a9a', 'Soma': '#9382a0', 'SNP': '#9a6c6c'}

def check_file(path):
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(path)

def safe_name(x):
    x = str(x)
    x = x.replace('|', '_')
    x = x.replace('/', '_')
    x = x.replace(' ', '_')
    return x

def all_combinations_ordered():
    combos = []
    for degree in [4, 3, 2, 1]:
        for comb in itertools.combinations(MODALITY_ORDER, degree):
            combos.append(tuple(comb))
    return combos

def get_modality_totals_from_counts(count_df, count_col='count'):
    totals = {}
    for modality in MODALITY_ORDER:
        total = 0
        for (_, row) in count_df.iterrows():
            comb = str(row['combination']).split('|')
            if modality in comb:
                total += int(row[count_col])
        totals[modality] = int(total)
    return totals

def print_modality_totals(title, totals):
    print('\n====================================')
    print(title)
    print('====================================')
    for modality in MODALITY_ORDER:
        print(f'{DISPLAY_LABELS[modality]:20s}: {int(totals[modality])}')

def load_risk_pids(endpoint, modality):
    risk_fp = BASE / RISK_MODELS[modality].format(endpoint=endpoint)
    check_file(risk_fp)
    df = pd.read_csv(str(OUTPUT_ROOT / risk_fp if (OUTPUT_ROOT / risk_fp).exists() else INPUT_ROOT / risk_fp))
    if 'ID' in df.columns:
        df = df.rename(columns={'ID': 'pid'})
    if 'pid' not in df.columns:
        raise ValueError(f'No pid column in risk file: {risk_fp}. Available columns: {list(df.columns)}')
    df['pid'] = pd.to_numeric(df['pid'], errors='coerce')
    df = df.dropna(subset=['pid']).copy()
    df['pid'] = df['pid'].astype(int)
    pids = set(df['pid'].dropna().astype(int).unique().tolist())
    print(f'[{endpoint} | {DISPLAY_LABELS[modality]}] original train samples:', len(pids))
    return pids

def build_modality_pid_sets(endpoint):
    modality_to_pids = {}
    for modality in MODALITY_ORDER:
        modality_to_pids[modality] = load_risk_pids(endpoint, modality)
    totals = {modality: len(modality_to_pids[modality]) for modality in MODALITY_ORDER}
    print_modality_totals('Original modality totals', totals)
    out_csv = OUT_DIR / f'CMCS_original_modality_totals.csv'
    pd.DataFrame({'modality': list(totals.keys()), 'display_label': [DISPLAY_LABELS[m] for m in totals.keys()], 'original_n_samples': list(totals.values())}).to_csv(out_csv, index=False)
    print('Saved original modality totals:', out_csv)
    return (modality_to_pids, totals)

def build_sample_presence(endpoint, modality_to_pids):
    all_pids = sorted(set.union(*[modality_to_pids[m] for m in MODALITY_ORDER]))
    rows = []
    for pid in all_pids:
        present = [m for m in MODALITY_ORDER if pid in modality_to_pids[m]]
        rows.append({'endpoint': endpoint, 'pid': pid, 'n_modalities': len(present), 'signature': '|'.join(present), **{m: int(m in present) for m in MODALITY_ORDER}})
    sample_df = pd.DataFrame(rows)
    out_csv = OUT_DIR / f'CMCS_train_sample_modality_presence_original.csv'
    sample_df.to_csv(out_csv, index=False)
    print('Saved original sample presence table:', out_csv)
    return sample_df

def build_exact_counts(sample_df, endpoint, suffix):
    rows = []
    for comb in all_combinations_ordered():
        comb_txt = '|'.join(comb)
        mask = np.ones(len(sample_df), dtype=bool)
        for modality in MODALITY_ORDER:
            if modality in comb:
                mask &= sample_df[modality].values == 1
            else:
                mask &= sample_df[modality].values == 0
        pids = sorted(sample_df.loc[mask, 'pid'].unique())
        count = len(pids)
        rows.append({'endpoint': endpoint, 'combination': comb_txt, 'display_combination': '|'.join([DISPLAY_LABELS[m] for m in comb]), 'degree': len(comb), 'count': count, 'pids': ';'.join([str(x) for x in pids])})
    count_df = pd.DataFrame(rows)
    out_csv = OUT_DIR / f'CMCS_train_sample_4set_venn_counts_exact_{suffix}.csv'
    count_df.to_csv(out_csv, index=False)
    print('Saved exact Venn counts:', out_csv)
    full_combo = '|'.join(MODALITY_ORDER)
    four_modal_n = int(count_df.loc[count_df['combination'] == full_combo, 'count'].sum())
    print(f'[{endpoint}] {suffix} four-modal overlap n = {four_modal_n}')
    return count_df

def save_exact_region_tables(endpoint, count_df, suffix):
    out_dir = REGION_DIR / suffix
    out_dir.mkdir(exist_ok=True, parents=True)
    summary_out = out_dir / f'CMCS_train_sample_4set_venn_exact_region_summary_{suffix}.csv'
    count_df.to_csv(summary_out, index=False)
    print('Saved exact region summary:', summary_out)
    for (_, row) in count_df.iterrows():
        combination = str(row['combination'])
        pids = str(row.get('pids', ''))
        if pids == '' or pids == 'nan':
            pid_list = []
        else:
            pid_list = [x for x in pids.split(';') if x != '']
        out_csv = out_dir / f'CMCS_{safe_name(combination)}_pids_{suffix}.csv'
        pd.DataFrame({'pid': pid_list}).to_csv(out_csv, index=False)

def plot_4set_venn_from_sets(endpoint, modality_sets, modality_totals, suffix, note):
    venn_input = OrderedDict()
    for modality in MODALITY_ORDER:
        label = f'{DISPLAY_LABELS[modality]}\nN = {int(modality_totals[modality])}'
        venn_input[label] = modality_sets[modality]
    (fig, ax) = plt.subplots(figsize=FIGSIZE)
    venn(venn_input, fmt='{size}', cmap=[VENN_COLORS[m] for m in MODALITY_ORDER], fontsize=3.8, legend_loc=None, ax=ax)
    ax.set_title(f'{ENDPOINT_LABEL} train-set sample overlap', fontsize=9, pad=6)
    ax.text(0.5, -0.035, note, transform=ax.transAxes, ha='center', va='top', fontsize=5.2)
    plt.subplots_adjust(left=0.04, right=0.96, top=0.9, bottom=0.1)
    out_pdf = OUT_DIR / f'CMCS_train_sample_4set_venn_{suffix}.pdf'
    out_svg = out_pdf.with_suffix('.svg')
    out_png = out_pdf.with_suffix('.png')
    plt.savefig(out_pdf, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.08)
    plt.savefig(out_svg, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.08)
    plt.savefig(out_png, dpi=600, transparent=False, bbox_inches='tight', pad_inches=0.08)
    plt.close()
    print('Saved 4-set Venn PDF:', out_pdf)
    print('Saved 4-set Venn SVG:', out_svg)
    print('Saved 4-set Venn PNG:', out_png)

def save_legend_figure(modality_totals, suffix):
    handles = [Patch(facecolor=VENN_COLORS[m], edgecolor='none', alpha=0.65, label=f'{DISPLAY_LABELS[m]} (N = {int(modality_totals[m])})') for m in MODALITY_ORDER]
    (fig, ax) = plt.subplots(figsize=LEGEND_FIGSIZE)
    ax.axis('off')
    ax.legend(handles=handles, loc='center', frameon=False, fontsize=8, title='Modalities', title_fontsize=9, ncol=2)
    out_pdf = OUT_DIR / f'CMCS_train_sample_4set_venn_legend_{suffix}.pdf'
    out_svg = out_pdf.with_suffix('.svg')
    out_png = out_pdf.with_suffix('.png')
    plt.savefig(out_pdf, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.05)
    plt.savefig(out_svg, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.05)
    plt.savefig(out_png, dpi=600, transparent=False, bbox_inches='tight', pad_inches=0.05)
    plt.close()
    print('Saved legend PDF:', out_pdf)
    print('Saved legend SVG:', out_svg)
    print('Saved legend PNG:', out_png)

def main():
    (modality_to_pids, totals) = build_modality_pid_sets(endpoint=ENDPOINT)
    sample_df = build_sample_presence(endpoint=ENDPOINT, modality_to_pids=modality_to_pids)
    counts = build_exact_counts(sample_df=sample_df, endpoint=ENDPOINT, suffix='observed')
    save_exact_region_tables(endpoint=ENDPOINT, count_df=counts, suffix='observed')
    plot_4set_venn_from_sets(endpoint=ENDPOINT, modality_sets=modality_to_pids, modality_totals=totals, suffix='observed', note='')
    save_legend_figure(modality_totals=totals, suffix='observed')
if __name__ == '__main__':
    main()
```

### Figure 1c — SG modality overlap


```python
import itertools
import logging
import warnings
from pathlib import Path
from collections import OrderedDict
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
try:
    from venn import venn
except ImportError:
    raise ImportError('Please install the venn package first:\npip install venn')
warnings.filterwarnings('ignore')
logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['svg.fonttype'] = 'none'
mpl.rcParams['font.family'] = 'DejaVu Sans'
BASE_DIR = Path('fig1_final/SingleModal_Run_printonly')
INPUT_FILES = {'SNP': BASE_DIR / 'snp_20251112_1617' / 'data' / 'test_pid_pool.csv', 'ECG': BASE_DIR / 'ecg_20251112_1620' / 'data' / 'test_pid_pool.csv', 'Soma': BASE_DIR / 'soma_20251112_1615' / 'data' / 'test_pid_pool.csv', 'Fundus': BASE_DIR / 'fundus_20251112_1621' / 'data' / 'test_pid_pool.csv'}
MODALITY_ORDER = ['SNP', 'ECG', 'Soma', 'Fundus']
DISPLAY_LABELS = {'SNP': 'SNP', 'ECG': 'ECG', 'Soma': 'Plasma proteomics', 'Fundus': 'Fundus'}
OUT_DIR = BASE_DIR / 'four_modal_test_pid_venn_forced_totals_ordered'
OUT_DIR.mkdir(exist_ok=True, parents=True)
REGION_DIR = OUT_DIR / 'exact_region_tables'
REGION_DIR.mkdir(exist_ok=True, parents=True)
MM_TO_INCH = 1 / 25.4
VENN_W_MM = 190
VENN_H_MM = 130
LEGEND_W_MM = 95
LEGEND_H_MM = 45
VENN_FIGSIZE = (VENN_W_MM * MM_TO_INCH, VENN_H_MM * MM_TO_INCH)
LEGEND_FIGSIZE = (LEGEND_W_MM * MM_TO_INCH, LEGEND_H_MM * MM_TO_INCH)
VENN_COLORS = {'Fundus': '#639ebe', 'ECG': '#1a6292', 'Soma': '#af7bac', 'SNP': '#7d5398'}
VENN_COLORS = {'Fundus': '#778368', 'ECG': '#658a9a', 'Soma': '#9382a0', 'SNP': '#9a6c6c'}

def check_file(path):
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f'File not found: {path}')

def safe_name(x):
    x = str(x)
    x = x.replace('|', '_')
    x = x.replace('/', '_')
    x = x.replace(' ', '_')
    return x

def all_combinations_ordered():
    combos = []
    for degree in [4, 3, 2, 1]:
        for comb in itertools.combinations(MODALITY_ORDER, degree):
            combos.append(tuple(comb))
    return combos

def get_modality_totals_from_counts(count_df, count_col='count'):
    totals = {}
    for modality in MODALITY_ORDER:
        total = 0
        for (_, row) in count_df.iterrows():
            comb = str(row['combination']).split('|')
            if modality in comb:
                total += int(row[count_col])
        totals[modality] = int(total)
    return totals

def print_totals(title, totals):
    print('\n====================================')
    print(title)
    print('====================================')
    for modality in MODALITY_ORDER:
        print(f'{DISPLAY_LABELS[modality]:20s}: {int(totals[modality])}')

def load_pid_set(csv_path, modality_name):
    check_file(csv_path)
    df = pd.read_csv(str(OUTPUT_ROOT / csv_path if (OUTPUT_ROOT / csv_path).exists() else INPUT_ROOT / csv_path))
    if 'pid' not in df.columns:
        raise ValueError(f"{modality_name} file must contain column 'pid'. Available columns: {list(df.columns)}")
    df['pid'] = pd.to_numeric(df['pid'], errors='coerce')
    df = df.dropna(subset=['pid']).copy()
    df['pid'] = df['pid'].astype(int)
    pid_set = set(df['pid'].unique().tolist())
    print(f'[{DISPLAY_LABELS[modality_name]}] loaded samples: {len(pid_set)}')
    print(f'[{DISPLAY_LABELS[modality_name]}] file: {csv_path}')
    return pid_set

def load_all_modality_sets():
    modality_to_pids = {}
    print('\n====================================')
    print('Loading test_pid_pool.csv')
    print('====================================')
    for modality in MODALITY_ORDER:
        modality_to_pids[modality] = load_pid_set(INPUT_FILES[modality], modality)
    original_totals = {m: len(modality_to_pids[m]) for m in MODALITY_ORDER}
    print_totals('Original modality totals', original_totals)
    out_csv = OUT_DIR / 'original_modality_totals.csv'
    pd.DataFrame({'modality': MODALITY_ORDER, 'display_label': [DISPLAY_LABELS[m] for m in MODALITY_ORDER], 'original_n_samples': [original_totals[m] for m in MODALITY_ORDER]}).to_csv(out_csv, index=False)
    print('Saved original modality totals:', out_csv)
    return (modality_to_pids, original_totals)

def build_sample_presence(modality_to_pids):
    all_pids = sorted(set.union(*[modality_to_pids[m] for m in MODALITY_ORDER]))
    rows = []
    for pid in all_pids:
        present = [m for m in MODALITY_ORDER if pid in modality_to_pids[m]]
        rows.append({'pid': pid, 'n_modalities': len(present), 'signature': '|'.join(present), **{m: int(m in present) for m in MODALITY_ORDER}})
    sample_df = pd.DataFrame(rows)
    out_csv = OUT_DIR / 'test_sample_modality_presence_original.csv'
    sample_df.to_csv(out_csv, index=False)
    print('Saved original sample presence table:', out_csv)
    return sample_df

def build_exact_counts(sample_df, suffix='original'):
    rows = []
    for comb in all_combinations_ordered():
        comb_txt = '|'.join(comb)
        mask = np.ones(len(sample_df), dtype=bool)
        for modality in MODALITY_ORDER:
            if modality in comb:
                mask &= sample_df[modality].values == 1
            else:
                mask &= sample_df[modality].values == 0
        pids = sorted(sample_df.loc[mask, 'pid'].dropna().astype(int).unique().tolist())
        rows.append({'combination': comb_txt, 'display_combination': '|'.join([DISPLAY_LABELS[m] for m in comb]), 'degree': len(comb), 'count': len(pids), 'pids': ';'.join([str(x) for x in pids])})
    count_df = pd.DataFrame(rows)
    out_csv = OUT_DIR / f'test_sample_4set_venn_exact_region_summary_{suffix}.csv'
    count_df.to_csv(out_csv, index=False)
    print('Saved exact region summary:', out_csv)
    return count_df

def save_exact_region_pid_tables(count_df, suffix):
    out_dir = REGION_DIR / suffix
    out_dir.mkdir(exist_ok=True, parents=True)
    for (_, row) in count_df.iterrows():
        combination = str(row['combination'])
        pids_txt = str(row.get('pids', ''))
        if pids_txt == '' or pids_txt == 'nan':
            pid_list = []
        else:
            pid_list = [int(float(x)) for x in pids_txt.split(';') if x != '']
        out_csv = out_dir / f'{safe_name(combination)}_pids_{suffix}.csv'
        pd.DataFrame({'pid': pid_list}).to_csv(out_csv, index=False)
    print('Saved region pid tables:', out_dir)

def plot_venn(modality_sets, modality_totals, suffix):
    venn_input = OrderedDict()
    for modality in MODALITY_ORDER:
        label = f'{DISPLAY_LABELS[modality]}\nN = {int(modality_totals[modality])}'
        venn_input[label] = modality_sets[modality]
    (fig, ax) = plt.subplots(figsize=VENN_FIGSIZE)
    venn(venn_input, fmt='{size}', cmap=[VENN_COLORS[m] for m in MODALITY_ORDER], fontsize=3.8, legend_loc=None, ax=ax)
    ax.set_title('Four-modal sample overlap', fontsize=10, pad=8)
    ax.text(0.5, -0.04, 'Layout: left-top SNP, right-top ECG, left-bottom Plasma proteomics, right-bottom Fundus. Forced totals are adjusted in single-modality regions only.', transform=ax.transAxes, ha='center', va='top', fontsize=5.2)
    plt.subplots_adjust(left=0.04, right=0.96, top=0.9, bottom=0.1)
    out_pdf = OUT_DIR / f'four_modal_test_pid_venn_{suffix}.pdf'
    out_svg = OUT_DIR / f'four_modal_test_pid_venn_{suffix}.svg'
    out_png = OUT_DIR / f'four_modal_test_pid_venn_{suffix}.png'
    plt.savefig(out_pdf, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.08)
    plt.savefig(out_svg, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.08)
    plt.savefig(out_png, dpi=600, transparent=False, bbox_inches='tight', pad_inches=0.08)
    plt.close()
    print('Saved venn PDF:', out_pdf)
    print('Saved venn SVG:', out_svg)
    print('Saved venn PNG:', out_png)

def save_legend_figure(modality_totals, suffix):
    handles = [Patch(facecolor=VENN_COLORS[m], edgecolor='none', alpha=0.65, label=f'{DISPLAY_LABELS[m]} (N = {int(modality_totals[m])})') for m in MODALITY_ORDER]
    (fig, ax) = plt.subplots(figsize=LEGEND_FIGSIZE)
    ax.axis('off')
    ax.legend(handles=handles, loc='center', frameon=False, fontsize=8, title='Modalities', title_fontsize=9, ncol=2)
    out_pdf = OUT_DIR / f'four_modal_test_pid_venn_legend_{suffix}.pdf'
    out_svg = OUT_DIR / f'four_modal_test_pid_venn_legend_{suffix}.svg'
    out_png = OUT_DIR / f'four_modal_test_pid_venn_legend_{suffix}.png'
    plt.savefig(out_pdf, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.05)
    plt.savefig(out_svg, dpi=300, transparent=True, bbox_inches='tight', pad_inches=0.05)
    plt.savefig(out_png, dpi=600, transparent=False, bbox_inches='tight', pad_inches=0.05)
    plt.close()
    print('Saved legend PDF:', out_pdf)
    print('Saved legend SVG:', out_svg)
    print('Saved legend PNG:', out_png)

def main():
    (modality_to_pids, totals) = load_all_modality_sets()
    sample_df = build_sample_presence(modality_to_pids)
    counts = build_exact_counts(sample_df, suffix='observed')
    save_exact_region_pid_tables(counts, suffix='observed')
    plot_venn(modality_sets=modality_to_pids, modality_totals=totals, suffix='observed')
    save_legend_figure(modality_totals=totals, suffix='observed')
if __name__ == '__main__':
    main()
```
