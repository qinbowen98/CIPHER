### Extended Data Figure 2b — Selected SNP genomic coverage


```python
import os
import re
import urllib.request
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.patches import Rectangle, Patch
from matplotlib.colors import Normalize, LinearSegmentedColormap
from matplotlib.cm import ScalarMappable
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['svg.fonttype'] = 'none'
mpl.rcParams['font.family'] = 'DejaVu Sans'
VARIANT_CSV = 'CHD_SNP_variant_gene_with_symbol.csv'
GWAS_CATALOG = 'gwas-catalog-download-associations-v1.0-full.tsv'
OUTDIR = 'figs2_final/snp_cytoband_coverage_preprocessing_onerow'
os.makedirs(OUTDIR, exist_ok=True)
CYTOBAND_FILE = os.path.join(OUTDIR, 'hg38_cytoBand.txt.gz')
OUTFIG = os.path.join(OUTDIR, 'prediction_SNP_cytoband_coverage_known_GWAS_loci_onerow.pdf')
OUTSVG = os.path.join(OUTDIR, 'prediction_SNP_cytoband_coverage_known_GWAS_loci_onerow.svg')
OUT_LEGEND_PDF = os.path.join(OUTDIR, 'prediction_SNP_cytoband_coverage_legend.pdf')
OUT_LEGEND_SVG = os.path.join(OUTDIR, 'prediction_SNP_cytoband_coverage_legend.svg')
OUT_COLORBAR = os.path.join(OUTDIR, 'prediction_SNP_density_colorbar.pdf')
OUT_COVERAGE_TABLE = os.path.join(OUTDIR, 'prediction_SNP_cytoband_coverage_known_GWAS_loci_summary.csv')
OUT_KNOWN_GWAS_TABLE = os.path.join(OUTDIR, 'selected_known_GWAS_hits_with_prediction_SNP_coverage.csv')
OUT_PRED_SNP_BAND_TABLE = os.path.join(OUTDIR, 'prediction_SNP_cytoband_counts.csv')
OUT_PRED_SNP_WITH_BAND = os.path.join(OUTDIR, 'prediction_SNPs_with_cytoband.csv')
CHR_ORDER = [f'chr{i}' for i in range(1, 23)] + ['chrX']
MANUAL_SELECTED_STUDIES = [{'endpoint': 'istroke', 'FIRST AUTHOR': 'Malik R', 'DATE': '2018-03-12', 'JOURNAL': 'Nat Genet', 'PUBMEDID': '29531354'}, {'endpoint': 'istroke', 'FIRST AUTHOR': 'Mishra A', 'DATE': '2022-09-30', 'JOURNAL': 'Nature', 'PUBMEDID': '36180795'}, {'endpoint': 'istroke', 'FIRST AUTHOR': 'Soderholm M', 'DATE': '2019-02-22', 'JOURNAL': 'Neurology', 'PUBMEDID': '30796134'}, {'endpoint': 'CHD', 'FIRST AUTHOR': 'Aragam KG', 'DATE': '2022-12-06', 'JOURNAL': 'Nat Genet', 'PUBMEDID': '36474045'}, {'endpoint': 'CHD', 'FIRST AUTHOR': 'Tcheandjieu C', 'DATE': '2022-08-01', 'JOURNAL': 'Nat Med', 'PUBMEDID': '35915156'}, {'endpoint': 'CHD', 'FIRST AUTHOR': 'Koyama S', 'DATE': '2020-10-05', 'JOURNAL': 'Nat Genet', 'PUBMEDID': '33020668'}, {'endpoint': 'CHD', 'FIRST AUTHOR': 'Lu X', 'DATE': '2012-07-01', 'JOURNAL': 'Nat Genet', 'PUBMEDID': '22751097'}, {'endpoint': 'CHD', 'FIRST AUTHOR': 'Nikpay M', 'DATE': '2015-09-07', 'JOURNAL': 'Nat Genet', 'PUBMEDID': '26343387'}]
CHD_PATTERNS = ['\\bcoronary heart disease\\b', '\\bcoronary artery disease\\b', '\\bischemic heart disease\\b', '\\bmyocardial infarction\\b', '\\bangina\\b', '\\bunstable angina\\b', '\\bacute coronary syndrome\\b', '\\batherosclerosis\\b', '\\bcoronary atherosclerosis\\b', '\\bcoronary stenosis\\b', '\\bchd\\b', '\\bcad\\b']
ISTROKE_PATTERNS = ['\\bischemic stroke\\b', '\\bischaemic stroke\\b', '\\bstroke\\b', '\\bcerebral infarction\\b', '\\bbrain infarction\\b', '\\blacunar stroke\\b', '\\bcardioembolic stroke\\b', '\\blarge artery stroke\\b', '\\bsmall vessel stroke\\b', '\\bsmall artery occlusion\\b', '\\btransient ischemic attack\\b', '\\btransient ischaemic attack\\b', '\\btia\\b']
FIGSIZE_MM = (200, 70)
FIGSIZE = (FIGSIZE_MM[0] / 25.4, FIGSIZE_MM[1] / 25.4)
CHROM_WIDTH = 0.16
SNP_TRACK_WIDTH = 0.055
SNP_TRACK_GAP = 0.035
COL_GAP = 0.43
TOP_PAD_MB = 14
BOTTOM_PAD_MB = 18
COLOR_CHROM = '#F2F2F2'
COLOR_CHROM_EDGE = '#9E9E9E'
COLOR_KNOWN_CHD = '#D55E5E'
COLOR_KNOWN_ISTROKE = '#5B8FD9'
COLOR_KNOWN_BOTH = '#9B59B6'
COLOR_COVERED = '#111111'
SNP_CMAP = LinearSegmentedColormap.from_list('snp_density_grey', ['#EEEEEE', '#111111'])

def download_cytoband_if_needed(path_out):
    if os.path.exists(str(OUTPUT_ROOT / path_out if (OUTPUT_ROOT / path_out).exists() else INPUT_ROOT / path_out)):
        return
    url = 'https://hgdownload.soe.ucsc.edu/goldenPath/hg38/database/cytoBand.txt.gz'
    print('Downloading cytoband file from:', url)
    urllib.request.urlretrieve(url, path_out)
    print('Saved:', path_out)

def normalize_text(x):
    if pd.isna(x):
        return ''
    return str(x).strip()

def normalize_date_to_yyyy_mm_dd(x):
    dt = pd.to_datetime(x, errors='coerce')
    if pd.isna(dt):
        return normalize_text(x)
    return dt.strftime('%Y-%m-%d')

def normalize_gwas_chr(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip()
    if s.upper() == 'X':
        return 'chrX'
    if s.upper() == 'Y':
        return 'chrY'
    if re.fullmatch('\\d+', s):
        val = int(s)
        if val == 23:
            return 'chrX'
        if val == 24:
            return 'chrY'
        return f'chr{val}'
    return np.nan

def parse_variant(v):
    m = re.match('(chr[0-9XY]+)_(\\d+)_', str(v))
    if m:
        return (m.group(1), int(m.group(2)))
    return (np.nan, np.nan)

def chrom_sort_key(chrom):
    x = str(chrom).replace('chr', '')
    if x == 'X':
        return 23
    if x == 'Y':
        return 24
    return int(x)

def build_regex(patterns):
    return re.compile('|'.join(patterns), flags=re.IGNORECASE)

def trait_match(series, regex_obj):
    return series.fillna('').astype(str).str.contains(regex_obj, regex=True, na=False)

def load_cytobands(path_in):
    bands = pd.read_csv(str(OUTPUT_ROOT / path_in if (OUTPUT_ROOT / path_in).exists() else INPUT_ROOT / path_in), sep='\t', compression='gzip', header=None, names=['chrom', 'start', 'end', 'band_name', 'gieStain'])
    bands = bands[bands['chrom'].isin(CHR_ORDER)].copy()
    bands = bands.sort_values(['chrom', 'start']).reset_index(drop=True)
    bands['band_id'] = bands['chrom'].astype(str) + ':' + bands['band_name'].astype(str) + ':' + bands['start'].astype(str) + '-' + bands['end'].astype(str)
    bands['cytoband'] = bands['chrom'].str.replace('chr', '', regex=False) + bands['band_name'].astype(str)
    bands['start_mb'] = bands['start'] / 1000000.0
    bands['end_mb'] = bands['end'] / 1000000.0
    bands['len_bp'] = bands['end'] - bands['start']
    bands['len_mb'] = bands['len_bp'] / 1000000.0
    return bands

def map_positions_to_bands(df, bands_df, chrom_col='chr', pos_col='pos'):
    if df.empty:
        return df.copy()
    out = []
    for (chrom, sub) in df.groupby(chrom_col):
        band_sub = bands_df[bands_df['chrom'] == chrom].sort_values('start').copy()
        if band_sub.empty:
            tmp = sub.copy()
            tmp['band_id'] = np.nan
            tmp['band_name'] = np.nan
            tmp['cytoband'] = np.nan
            out.append(tmp)
            continue
        starts = band_sub['start'].to_numpy()
        ends = band_sub['end'].to_numpy()
        band_ids = band_sub['band_id'].to_numpy()
        band_names = band_sub['band_name'].to_numpy()
        cytobands = band_sub['cytoband'].to_numpy()
        pos = sub[pos_col].to_numpy()
        idx = np.searchsorted(starts, pos, side='right') - 1
        band_id_arr = np.array([np.nan] * len(sub), dtype=object)
        band_name_arr = np.array([np.nan] * len(sub), dtype=object)
        cytoband_arr = np.array([np.nan] * len(sub), dtype=object)
        valid = idx >= 0
        valid_idx = idx[valid]
        valid_pos = pos[valid]
        inside = valid_pos < ends[valid_idx]
        pos_valid_indices = np.where(valid)[0][inside]
        band_valid_indices = valid_idx[inside]
        band_id_arr[pos_valid_indices] = band_ids[band_valid_indices]
        band_name_arr[pos_valid_indices] = band_names[band_valid_indices]
        cytoband_arr[pos_valid_indices] = cytobands[band_valid_indices]
        tmp = sub.copy()
        tmp['band_id'] = band_id_arr
        tmp['band_name'] = band_name_arr
        tmp['cytoband'] = cytoband_arr
        out.append(tmp)
    return pd.concat(out, ignore_index=True)

def apply_manual_study_flags_all_hits(gwas_df, manual_studies):
    gwas_df = gwas_df.copy()
    gwas_df['is_chd_related'] = False
    gwas_df['is_istroke_related'] = False
    gwas_df['manual_study_selected'] = False
    for col in ['FIRST AUTHOR', 'JOURNAL', 'PUBMEDID']:
        if col in gwas_df.columns:
            gwas_df[f'__{col}'] = gwas_df[col].apply(normalize_text)
        else:
            gwas_df[f'__{col}'] = ''
    if 'DATE' not in gwas_df.columns:
        raise ValueError('GWAS catalog does not contain DATE column.')
    gwas_df['__DATE'] = gwas_df['DATE'].apply(normalize_date_to_yyyy_mm_dd)
    trait_series = gwas_df['DISEASE/TRAIT'].fillna('').astype(str)
    study_series = gwas_df.get('STUDY', pd.Series([''] * len(gwas_df), index=gwas_df.index)).fillna('').astype(str)
    chd_regex = build_regex(CHD_PATTERNS)
    istroke_regex = build_regex(ISTROKE_PATTERNS)
    chd_trait_mask = trait_match(trait_series, chd_regex) | trait_match(study_series, chd_regex)
    istroke_trait_mask = trait_match(trait_series, istroke_regex) | trait_match(study_series, istroke_regex)
    for item in manual_studies:
        endpoint = item['endpoint']
        author = normalize_text(item['FIRST AUTHOR'])
        date = normalize_date_to_yyyy_mm_dd(item['DATE'])
        journal = normalize_text(item['JOURNAL'])
        pubmedid = normalize_text(item['PUBMEDID'])
        study_mask = (gwas_df['__FIRST AUTHOR'] == author) & (gwas_df['__DATE'] == date) & (gwas_df['__JOURNAL'] == journal) & (gwas_df['__PUBMEDID'] == pubmedid)
        if endpoint == 'CHD':
            final_mask = study_mask & chd_trait_mask
            gwas_df.loc[final_mask, 'is_chd_related'] = True
        elif endpoint == 'istroke':
            final_mask = study_mask & istroke_trait_mask
            gwas_df.loc[final_mask, 'is_istroke_related'] = True
        else:
            raise ValueError(f'Unknown endpoint: {endpoint}')
        gwas_df.loc[study_mask, 'manual_study_selected'] = True
        print(f'[Manual study] {endpoint} | {author} {date} {journal} PMID={pubmedid} | all hits in study = {int(study_mask.sum())} | endpoint-matched hits = {int(final_mask.sum())}')
    helper_cols = [c for c in gwas_df.columns if c.startswith('__')]
    gwas_df = gwas_df.drop(columns=helper_cols)
    return gwas_df

def known_region_type(row):
    if row['known_chd_hits'] > 0 and row['known_istroke_hits'] > 0:
        return 'both'
    if row['known_chd_hits'] > 0:
        return 'chd'
    if row['known_istroke_hits'] > 0:
        return 'istroke'
    return 'none'

def known_region_color(region_type):
    if region_type == 'chd':
        return COLOR_KNOWN_CHD
    if region_type == 'istroke':
        return COLOR_KNOWN_ISTROKE
    if region_type == 'both':
        return COLOR_KNOWN_BOTH
    return COLOR_CHROM

def known_region_fill(region_type):
    if region_type == 'chd':
        return '#F5CACA'
    if region_type == 'istroke':
        return '#CFE0F6'
    if region_type == 'both':
        return '#E6D4F0'
    return COLOR_CHROM
download_cytoband_if_needed(CYTOBAND_FILE)
bands = load_cytobands(CYTOBAND_FILE)
chrom_sizes = bands.groupby('chrom', as_index=False)['end_mb'].max().rename(columns={'end_mb': 'chrom_len_mb'})
chrom_len_dict = dict(zip(chrom_sizes['chrom'], chrom_sizes['chrom_len_mb']))
variant = pd.read_csv(str(OUTPUT_ROOT / VARIANT_CSV if (OUTPUT_ROOT / VARIANT_CSV).exists() else INPUT_ROOT / VARIANT_CSV))
if 'variant_id' not in variant.columns:
    raise ValueError("VARIANT_CSV must contain 'variant_id' column.")
if 'SNP' not in variant.columns:
    raise ValueError("VARIANT_CSV must contain 'SNP' column.")
variant[['chr', 'pos']] = variant['variant_id'].apply(lambda x: pd.Series(parse_variant(x)))
variant = variant.dropna(subset=['chr', 'pos']).copy()
variant['pos'] = variant['pos'].astype(int)
variant = variant[variant['chr'].isin(CHR_ORDER)].copy()
pred_snp_band = map_positions_to_bands(variant, bands, chrom_col='chr', pos_col='pos')
pred_snp_band = pred_snp_band.dropna(subset=['band_id']).copy()
pred_band_counts = pred_snp_band.groupby('band_id', as_index=False).agg(n_prediction_snps=('SNP', 'nunique'), n_prediction_rows=('SNP', 'size'))
pred_snp_band.to_csv(OUT_PRED_SNP_WITH_BAND, index=False)
gwas = pd.read_csv(str(OUTPUT_ROOT / GWAS_CATALOG if (OUTPUT_ROOT / GWAS_CATALOG).exists() else INPUT_ROOT / GWAS_CATALOG), sep='\t', low_memory=False)
gwas.columns = gwas.columns.str.strip()
if 'CHR_ID' not in gwas.columns or 'CHR_POS' not in gwas.columns:
    raise ValueError('GWAS catalog must contain CHR_ID and CHR_POS columns.')
gwas['CHR_POS'] = pd.to_numeric(gwas['CHR_POS'], errors='coerce')
gwas = gwas.dropna(subset=['CHR_ID', 'CHR_POS']).copy()
gwas['chr'] = gwas['CHR_ID'].apply(normalize_gwas_chr)
gwas = gwas[gwas['chr'].isin(CHR_ORDER)].copy()
gwas['pos'] = gwas['CHR_POS'].astype(int)
print('\nGWAS chromosome counts after normalization:')
print(gwas['chr'].value_counts().reindex(CHR_ORDER).fillna(0).astype(int).to_string())
gwas_flagged = apply_manual_study_flags_all_hits(gwas_df=gwas, manual_studies=MANUAL_SELECTED_STUDIES)
known_hits = gwas_flagged[gwas_flagged['is_chd_related'] | gwas_flagged['is_istroke_related']].copy()
known_hits_mapped = map_positions_to_bands(known_hits, bands, chrom_col='chr', pos_col='pos')
known_hits_mapped = known_hits_mapped.dropna(subset=['band_id']).copy()
band_table = bands[['band_id', 'chrom', 'band_name', 'cytoband', 'start', 'end', 'start_mb', 'end_mb', 'len_bp', 'len_mb']].copy()
band_table = band_table.merge(pred_band_counts, on='band_id', how='left')
band_table['n_prediction_snps'] = band_table['n_prediction_snps'].fillna(0).astype(int)
band_table['n_prediction_rows'] = band_table['n_prediction_rows'].fillna(0).astype(int)
band_table['has_prediction_snp'] = band_table['n_prediction_snps'] > 0
known_all_counts = known_hits_mapped.groupby('band_id', as_index=False).size().rename(columns={'size': 'known_hits_total'})
known_chd_counts = known_hits_mapped[known_hits_mapped['is_chd_related']].groupby('band_id', as_index=False).size().rename(columns={'size': 'known_chd_hits'})
known_istroke_counts = known_hits_mapped[known_hits_mapped['is_istroke_related']].groupby('band_id', as_index=False).size().rename(columns={'size': 'known_istroke_hits'})
band_table = band_table.merge(known_all_counts, on='band_id', how='left')
band_table = band_table.merge(known_chd_counts, on='band_id', how='left')
band_table = band_table.merge(known_istroke_counts, on='band_id', how='left')
for col in ['known_hits_total', 'known_chd_hits', 'known_istroke_hits']:
    band_table[col] = band_table[col].fillna(0).astype(int)
band_table['known_region_type'] = band_table.apply(known_region_type, axis=1)
band_table['known_region_has_prediction_snp'] = (band_table['known_hits_total'] > 0) & band_table['has_prediction_snp']
band_table.to_csv(OUT_PRED_SNP_BAND_TABLE, index=False)
known_hits_mapped = known_hits_mapped.merge(band_table[['band_id', 'has_prediction_snp', 'n_prediction_snps']], on='band_id', how='left')
known_hits_mapped['covered_by_prediction_snp_cytoband'] = known_hits_mapped['has_prediction_snp'].fillna(False)
known_hits_mapped.to_csv(OUT_KNOWN_GWAS_TABLE, index=False)
n_known_hits = len(known_hits_mapped)
n_known_hits_covered = int(known_hits_mapped['covered_by_prediction_snp_cytoband'].sum())
coverage_ratio = n_known_hits_covered / n_known_hits if n_known_hits > 0 else np.nan
summary_rows = []
for (endpoint, col) in [('CHD', 'is_chd_related'), ('istroke', 'is_istroke_related')]:
    sub = known_hits_mapped[known_hits_mapped[col]].copy()
    n = len(sub)
    n_cov = int(sub['covered_by_prediction_snp_cytoband'].sum())
    summary_rows.append({'endpoint': endpoint, 'known_GWAS_hits_mapped_to_cytoband': n, 'known_GWAS_hits_covered_by_prediction_SNP_cytoband': n_cov, 'coverage_ratio': n_cov / n if n > 0 else np.nan})
summary_rows.append({'endpoint': 'CHD_or_istroke', 'known_GWAS_hits_mapped_to_cytoband': n_known_hits, 'known_GWAS_hits_covered_by_prediction_SNP_cytoband': n_known_hits_covered, 'coverage_ratio': coverage_ratio})
coverage_summary = pd.DataFrame(summary_rows)
coverage_summary.to_csv(OUT_COVERAGE_TABLE, index=False)
print('\n==============================')
print('Prediction SNP cytoband coverage of known GWAS loci')
print('==============================')
print(coverage_summary.to_string(index=False))
print(f'\nMain coverage number: {n_known_hits_covered:,} / {n_known_hits:,} = {coverage_ratio:.2%}')
plot_chroms = [c for c in CHR_ORDER if c in chrom_len_dict]
max_chr_len = max([chrom_len_dict[c] for c in plot_chroms])
x_positions = {chrom: i * COL_GAP for (i, chrom) in enumerate(plot_chroms)}
snp_values = band_table['n_prediction_snps'].to_numpy()
snp_values_nonzero = snp_values[snp_values > 0]
if len(snp_values_nonzero) > 0:
    vmax = np.quantile(snp_values_nonzero, 0.99)
    if vmax <= 0:
        vmax = snp_values_nonzero.max()
else:
    vmax = 1
norm = Normalize(vmin=0, vmax=vmax)
(fig, ax) = plt.subplots(figsize=FIGSIZE)
for chrom in plot_chroms:
    x0 = x_positions[chrom]
    y_base = 0
    sub = band_table[band_table['chrom'] == chrom].sort_values('start').copy()
    chrom_x = x0
    snp_track_x = chrom_x + CHROM_WIDTH + SNP_TRACK_GAP
    chrom_len = chrom_len_dict[chrom]
    for (_, row) in sub.iterrows():
        y = y_base + row['start_mb']
        h = row['len_mb']
        region_type = row['known_region_type']
        face = known_region_fill(region_type)
        ax.add_patch(Rectangle((chrom_x, y), CHROM_WIDTH, h, facecolor=face, edgecolor='white', linewidth=0.08, zorder=1))
        if row['known_hits_total'] > 0:
            edge_color = known_region_color(region_type)
            ax.add_patch(Rectangle((chrom_x - 0.01, y), CHROM_WIDTH + 0.02, h, facecolor='none', edgecolor=edge_color, linewidth=0.5, zorder=4))
    ax.add_patch(Rectangle((chrom_x, y_base), CHROM_WIDTH, chrom_len, facecolor='none', edgecolor=COLOR_CHROM_EDGE, linewidth=0.55, zorder=5))
    for (_, row) in sub.iterrows():
        y = y_base + row['start_mb']
        h = row['len_mb']
        n_snp = row['n_prediction_snps']
        if n_snp > 0:
            color = SNP_CMAP(norm(n_snp))
        else:
            color = '#FFFFFF'
        ax.add_patch(Rectangle((snp_track_x, y), SNP_TRACK_WIDTH, h, facecolor=color, edgecolor='white', linewidth=0.05, zorder=2))
    ax.add_patch(Rectangle((snp_track_x, y_base), SNP_TRACK_WIDTH, chrom_len, facecolor='none', edgecolor=COLOR_CHROM_EDGE, linewidth=0.45, zorder=5))
    ax.text(chrom_x + CHROM_WIDTH / 2, y_base - 7, chrom.replace('chr', ''), ha='center', va='bottom', fontsize=6.5)
coverage_text = f'Known GWAS loci covered by prediction SNP cytobands: {n_known_hits_covered:,}/{n_known_hits:,} ({coverage_ratio:.1%})'
ax.text(0.01, 0.985, coverage_text, transform=ax.transAxes, ha='left', va='top', fontsize=7.5)
x_max = (len(plot_chroms) - 1) * COL_GAP + CHROM_WIDTH + SNP_TRACK_GAP + SNP_TRACK_WIDTH + 0.3
y_min = -TOP_PAD_MB
y_max = max_chr_len + BOTTOM_PAD_MB
ax.set_xlim(-0.2, x_max)
ax.set_ylim(y_max, y_min)
ax.set_xticks([])
ax.set_yticks([])
for spine in ['top', 'right', 'left', 'bottom']:
    ax.spines[spine].set_visible(False)
plt.tight_layout()
plt.savefig(OUTFIG, dpi=300, transparent=True, bbox_inches='tight')
plt.savefig(OUTSVG, dpi=300, transparent=True, bbox_inches='tight')
plt.close()
print('\nSaved main figure:')
print(OUTFIG)
print(OUTSVG)
legend_handles = [Patch(facecolor='#F5CACA', edgecolor=COLOR_KNOWN_CHD, label='Known CHD cytoband'), Patch(facecolor='#CFE0F6', edgecolor=COLOR_KNOWN_ISTROKE, label='Known iStroke cytoband'), Patch(facecolor='#E6D4F0', edgecolor=COLOR_KNOWN_BOTH, label='Shared known cytoband'), Patch(facecolor=COLOR_COVERED, edgecolor='none', label='Covered by prediction SNPs'), Patch(facecolor='#FFFFFF', edgecolor=COLOR_CHROM_EDGE, label='Prediction SNP density track')]
(fig_leg, ax_leg) = plt.subplots(figsize=(3.2, 1.4))
ax_leg.axis('off')
ax_leg.legend(handles=legend_handles, frameon=False, fontsize=8, loc='center', ncol=1)
plt.tight_layout()
plt.savefig(OUT_LEGEND_PDF, dpi=300, transparent=True, bbox_inches='tight')
plt.savefig(OUT_LEGEND_SVG, dpi=300, transparent=True, bbox_inches='tight')
plt.close(fig_leg)
print('\nSaved separate legend:')
print(OUT_LEGEND_PDF)
print(OUT_LEGEND_SVG)
(fig_cb, ax_cb) = plt.subplots(figsize=(1.0, 2.6))
sm = ScalarMappable(norm=norm, cmap=SNP_CMAP)
sm.set_array([])
cbar = plt.colorbar(sm, cax=ax_cb, orientation='vertical')
cb_ticks = np.linspace(0, vmax, 3)
cbar.set_ticks(cb_ticks)
cbar.set_ticklabels([f'{x:.0f}' for x in cb_ticks])
cbar.set_label('Prediction SNPs per cytoband', fontsize=8)
cbar.ax.tick_params(labelsize=7)
plt.savefig(OUT_COLORBAR, dpi=300, transparent=True, bbox_inches='tight')
plt.close(fig_cb)
print('\nSaved SNP density colorbar:')
print(OUT_COLORBAR)
print('\nTop cytobands by prediction SNP count:')
print(band_table.sort_values('n_prediction_snps', ascending=False)[['chrom', 'band_name', 'cytoband', 'n_prediction_snps', 'known_hits_total', 'known_chd_hits', 'known_istroke_hits', 'has_prediction_snp']].head(30).to_string(index=False))
print('\nKnown GWAS cytobands not covered by prediction SNPs:')
not_covered = band_table[(band_table['known_hits_total'] > 0) & ~band_table['has_prediction_snp']].copy()
if not_covered.empty:
    print('None. All known GWAS cytobands are covered by at least one prediction SNP.')
else:
    print(not_covered[['chrom', 'band_name', 'cytoband', 'known_hits_total', 'known_chd_hits', 'known_istroke_hits', 'n_prediction_snps']].to_string(index=False))
```

### Supporting analysis for Extended Data Figure 2d — Protein-panel GO enrichment


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gseapy as gp
SVM_FILE = 'CIPHER_save_11_18/SingleModal_Run_update_v2/soma_20251117_1322/explanations/CHD_soma_TR_best_IG.csv'
NCR_FILE = 'CIPHER_save_11_18/SingleModal_Run_update_v2/soma_20251117_1324/explanations/CHD_soma_TR_best_IG.csv'
CELL_FILE = 'code_02_02_save/soma_20260202_2006/explanations/stroke_soma_TR_best_IG.csv'
OUT_DIR = 'GO_enrichment_final'
os.makedirs(OUT_DIR, exist_ok=True)

def parse_protein(x):
    if pd.isna(x):
        return None
    x = str(x)
    if '.' in x:
        x = x.split('.')[1]
    return x

def load_protein_list(path):
    df = pd.read_csv(str(OUTPUT_ROOT / path if (OUTPUT_ROOT / path).exists() else INPUT_ROOT / path))
    if 'feature' not in df.columns:
        raise ValueError(f'{path} missing feature column')
    genes = df['feature'].dropna().map(parse_protein)
    genes = genes.dropna().unique().tolist()
    return genes

def run_go_enrichr(gene_list, out_prefix):
    gene_list = list(set(gene_list))
    print(f'\n[GO] Running: {out_prefix}')
    print(f'[GO] Genes: {len(gene_list)}')
    print(f'[GO] Example: {gene_list[:10]}')
    enr = gp.enrichr(gene_list=gene_list, gene_sets=['GO_Biological_Process_2023', 'GO_Molecular_Function_2023', 'GO_Cellular_Component_2023'], organism='Human', outdir=out_prefix, cutoff=0.5, no_plot=True)
    res = enr.results.copy()
    res.to_csv(f'{out_prefix}_GO_full.csv', index=False)
    return res

def plot_go(df, title, out_png):
    if df is None or len(df) == 0:
        print(f'[Plot] empty: {title}')
        return
    sig = df[df['Adjusted P-value'] < 0.05].copy()
    if len(sig) == 0:
        print(f'[Plot] no significant GO: {title}')
        return
    sig = sig.sort_values('Adjusted P-value').head(15)
    sig['score'] = -np.log10(sig['Adjusted P-value'] + 1e-300)
    plt.figure(figsize=(6, 5))
    plt.barh(sig['Term'][::-1], sig['score'][::-1])
    plt.xlabel('-log10 FDR')
    plt.title(title)
    plt.tight_layout()
    plt.savefig(out_png, dpi=300, bbox_inches='tight', transparent=True)
    plt.show()

def run_set(name, path):
    proteins = load_protein_list(path)
    out_prefix = os.path.join(OUT_DIR, name)
    res = run_go_enrichr(proteins, out_prefix)
    plot_go(res, f'{name} GO enrichment', os.path.join(OUT_DIR, f'{name}_GO.png'))
    return res

def main():
    print('\n==============================')
    print('GO enrichment pipeline')
    print('==============================\n')
    run_set('SVM_select', SVM_FILE)
    run_set('NCR_select', NCR_FILE)
    run_set('cell_select', CELL_FILE)
    print('\nDONE')
    print('Output:', OUT_DIR)
if __name__ == '__main__':
    main()
```

### Supporting analysis for Extended Data Figure 2d — Protein-panel overlap


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
SVM_FILE = 'CIPHER_save_11_18/SingleModal_Run_update_v2/soma_20251117_1322/explanations/CHD_soma_TR_best_IG.csv'
NCR_FILE = 'CIPHER_save_11_18/SingleModal_Run_update_v2/soma_20251117_1324/explanations/CHD_soma_TR_best_IG.csv'
CELL_FILE = 'code_02_02_save/soma_20260202_2006/explanations/stroke_soma_TR_best_IG.csv'
CAUSAL_GENES = ['PCSK9', 'NOS3', 'LDLR', 'LPL', 'LIPA', 'ANGPTL4', 'CDK8', 'APOB', 'FN1', 'CCM2', 'HNF1A', 'PNPLA3', 'REST', 'SCARB1', 'APOE', 'ABCG8', 'SEMA3C', 'COL4A1', 'MECOM', 'SGCD', 'ACVRL1', 'RRP1B', 'SWAP7', 'IL6R', 'NME7', 'LAMB2', 'ARHGEF26', 'EDNRA', 'GUCY1A3', 'LOX', 'PHACTR1', 'TCF21', 'PLG', 'TBX20', 'ARHGAP42', 'APOC3', 'LIPC', 'MYH11', 'CETP', 'SREBF1', 'CXCR4', 'MC4R', 'TGFB1', 'FES']

def parse_protein(x):
    if pd.isna(x):
        return None
    x = str(x)
    if '.' in x:
        x = x.split('.')[1]
    return x

def load_set(path):
    df = pd.read_csv(str(OUTPUT_ROOT / path if (OUTPUT_ROOT / path).exists() else INPUT_ROOT / path))
    genes = df['feature'].dropna().map(parse_protein)
    genes = set([g for g in genes if g is not None])
    return genes

def compute_ratio(selected, reference):
    overlap = selected.intersection(reference)
    ratio = len(overlap) / len(selected) if len(selected) > 0 else 0
    return ratio

def main():
    svm = load_set(SVM_FILE)
    ncr = load_set(NCR_FILE)
    cell = load_set(CELL_FILE)
    ref = set(CAUSAL_GENES)
    ratios = {'SVM_select': compute_ratio(svm, ref), 'NCR_select': compute_ratio(ncr, ref), 'Cell_select': compute_ratio(cell, ref)}
    print('\nOverlap ratio results:')
    for (k, v) in ratios.items():
        print(k, round(v, 4))
    labels = list(ratios.keys())
    values = list(ratios.values())
    plt.figure(figsize=(5.5, 4))
    bars = plt.bar(labels, values)
    plt.ylabel('Causal gene overlap ratio')
    plt.title('Protein selection enrichment')
    for (i, v) in enumerate(values):
        plt.text(i, v + 0.005, f'{v:.3f}', ha='center')
    plt.ylim(0, max(values) * 1.3 if max(values) > 0 else 1)
    plt.tight_layout()
    out = 'gene_overlap_ratio_barplot.png'
    plt.savefig(out, dpi=300, bbox_inches='tight')
    plt.show()
    print('\nSaved:', out)
if __name__ == '__main__':
    main()
```
